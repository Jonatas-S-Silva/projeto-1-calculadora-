**CALCULADORA**


```python
while True:
    # O menu agora é impresso no início de cada repetição
    print('\n----------   CALCULADORA   ----------\n')
    print('             -> MENU\n')
    print('1- ADÇÃO      2- SUBTRAÇÃO   3- MULTIPLICAÇÃO')
    print('4- DIVISÃO    5- POTÊNCIA    6- PORCENTAGEM')

    try:
        opcao = input('\nSelecione uma opção (ou digite "sair" para encerrar): ').strip().lower()
        if opcao == 'sair':
            print('Encerrando a calculadora. Até mais!')
            break

        opcao = int(opcao)

        if opcao < 1 or opcao > 6:
            print('Opção inválida! Escolha um número de 1 a 6 ou digite "sair".')
            continue

        # ADÇÃO
        if opcao == 1:
            soma1 = input('Digite primeiro número:')
            soma2 = input('Digite segundo número:')
            soma1 = float(soma1.replace(',', '.'))
            soma2 = float(soma2.replace(',', '.'))
            soma = soma1 + soma2
            print('A soma de ', soma1, '+', soma2, '=', soma)

        # SUBTRAÇÃO
        elif opcao == 2:
            sub1 = input('Digite primeiro número:')
            sub2 = input('Digite segundo número:')
            sub1 = float(sub1.replace(',', '.'))
            sub2 = float(sub2.replace(',', '.'))
            sub = sub1 - sub2
            print('A subtração de ', sub1, '-', sub2, '=', sub)

        # MULTIPLICAÇÃO
        elif opcao == 3:
            mut1 = input('Digite primeiro número:')
            mut2 = input('Digite segundo número:')
            mut1 = float(mut1.replace(',', '.'))
            mut2 = float(mut2.replace(',', '.'))
            mut = mut1 * mut2
            print('A multiplicação de ', mut1, '*', mut2, '=', mut)

        # DIVISÃO
        elif opcao == 4:
            divisao1 = input('Digite primeiro número:')
            divisao2 = input('Digite segundo número:')
            divisao1 = float(divisao1.replace(',', '.'))
            divisao2 = float(divisao2.replace(',', '.'))
            if divisao2 == 0:
                print('Erro: Divisão por zero não é permitida!')
                continue
            divisao = divisao1 / divisao2
            print('A divisão de ', divisao1, '/', divisao2, '=', divisao)

        # POTÊNCIA
        elif opcao == 5:
            potencia1 = input('Digite primeiro número (base):')
            potencia2 = input('Digite segundo número (expoente):')
            potencia1 = float(potencia1.replace(',', '.'))
            potencia2 = float(potencia2.replace(',', '.'))
            potencia = potencia1 ** potencia2
            print('A potência de base ', potencia1, 'elevado a ', potencia2, '= ', potencia)

        # PORCENTAGEM
        elif opcao == 6:
            porc_1 = input('Digite a porcetagem (%):')
            porc_2 = input('Digite o valor que vai extrair a porcentagem:')
            porc_1 = float(porc_1.replace(',', '.'))
            porc_2 = float(porc_2.replace(',', '.'))
            porc = (porc_1) * (porc_2) / (100)
            print('A porcentagem de ', porc_1, '% de ', porc_2, ' = ', porc)

    except ValueError:
        print('Erro: Por favor, digite apenas números válidos!')
```
